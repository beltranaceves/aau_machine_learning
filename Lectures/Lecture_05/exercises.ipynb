## Exercise 2 

Consider a poker game consisting of two rounds, and where each player is initially dealt three cards. During the first round all three cards can be changed (FC), but during the second round at most two cards can be changed (SC). When deciding on whether to call or fold you can take into account the number of cards changed by your opponent as well as your current hand (MH). After playing 20 games we have the results in Table 1, where BH shows who has the best hand.

```python
df = pd.read_csv('poker_data.csv')
print(df)
```

* Construct a naive Bayes classifier for the poker domain. 
* Use the data cases to learn the maximum likelihood parameters in the model; if you feel comfortable with the estimation procedure, you only need to estimate the probabilities required for solving the exercise below. 
* What is the posterior distribution of BH given MH=1a, FC=1, and SC=1? 
* Do you see any potential issues with the posterior and how could you address it?

```python
# Hint: You can calculate counts using, e.g.:
df[(df.FC==1) & (df.BH=='me')].shape[0]

from collections import defaultdict

# Store all conditional probabilities
probs = defaultdict(dict)

# Calculate P(BH)
P_BH = df['BH'].value_counts() / len(df)

# For each feature
for feature in ['MH', 'FC', 'SC']:
    for bh_value in ['me', 'op', 'draw']:
        # P(feature_value | BH)
        subset = df[df['BH'] == bh_value]
        probs[f'{feature}|{bh_value}'] = subset[feature].value_counts(normalize=True)

# Prediction: query = {'MH': '1a', 'FC': 1, 'SC': 1}
query = {'MH': '1a', 'FC': 1, 'SC': 1}

posteriors = {}
for bh_class in ['me', 'op', 'draw']:
    score = P_BH[bh_class]
    for feature, value in query.items():
        key = f'{feature}|{bh_class}'
        score *= probs[key].get(value, 0.0)  # Returns 0 if not found coz errors and shiet
    posteriors[bh_class] = score

# Normalize to get probabilities
total = sum(posteriors.values())
for bh_class in posteriors:
    posteriors[bh_class] /= total
print("Posterior probabilities:", posteriors)
```
